```r
title = "proportion developed"
w <- 2.5
h <- 2
options(repr.plot.width=w, repr.plot.height=h)
  
p = ggplot(df_sample, aes(x = condition , y = proportion_developed, fill  = condition2)) +  # dots for each file
    stat_summary(
      fun = mean, 
      geom = "bar",
      position = position_dodge(width = 0.75), alpha = 0.8, width = 0.7, fill = "grey") +   # error bars
    geom_jitter(
      aes(fill = condition2, color = condition2),
      position = position_jitterdodge(jitter.width = 0.15, dodge.width = 0.75),
      size = 0.5, alpha = 0.9
    )  +  # average bar
    stat_summary(
      fun.data = mean_se, 
      geom = "errorbar",
      position = position_dodge(width = 0.75),
      width = 0.2, 
      color = "#505150")+
    labs(
      title = title,
      y = "proportion developed",
      x = "condition"
    )+ settheme+
    theme(
      axis.text.x = element_text(angle = 45, hjust = 1),
      #legend.position = "none", 
      legend.key.size = unit(0.3, "cm")
      
    ) +
      scale_y_continuous(limits = c(0, 0.7), expand = c(0, 0))+
      #facet_wrap( ~ condition_2) +
      scale_color_manual(values=col_condition2)

ggsave(file.path(out_dir, sprintf("A_%s.pdf", title)),
                plot = p, width = w, height = h)

p
```

```r
df_exp <- df_sample %>%
  group_by(condition, condition2) %>%
  summarise(
    developed = sum(developed),
    failed = sum(failed),
    proportion_developed = developed/(developed+failed),
    .groups = "drop"
  )

df_exp 
```

```r
count_matrix <- as.matrix(
  df_exp[, c("developed", "failed")]
)

rownames(count_matrix) <-df_exp$condition

overall_chisq <- chisq.test(count_matrix)

overall_chisq
overall_chisq$expected
```

```r
library(dplyr)
library(purrr)
library(tibble)

control_name <- "H90.5: H9R0.5"

control_counts <- df_exp%>%
  filter(condition == control_name) %>%
  select(developed, failed)

fisher_results <- df_exp %>%
  filter(condition != control_name) %>%
  mutate(
    fisher = map2(
      developed,
      failed,
      ~ fisher.test(
          matrix(
            c(
              control_counts$developed,
              control_counts$failed,
              .x,
              .y
            ),
            nrow = 2,
            byrow = TRUE
          ),
          alternative = "two.sided"
        )
    ),
    odds_ratio = map_dbl(fisher, ~ unname(.x$estimate)),
    p_value = map_dbl(fisher, ~ .x$p.value),
    p_adj = p.adjust(p_value, method = "holm")
  ) %>%
  select(
    condition,
    developed,
    failed,
    proportion_developed,
    odds_ratio,
    p_value,
    p_adj
  )

fisher_results
```
